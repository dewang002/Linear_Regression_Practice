```python
df["year"] = pd.to_numeric(df["year"], errors="coerce")
```

```python
df['year'].isnull().sum()/ len(df['year']) * 100
```

```python
df['year'] = df['year'].dropna()
```

```python
df['year'] = pd.to_datetime(df['year'])
```

```python
df.info()
```

```python
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace(r"[^\d]", "", regex=True)) 
```

```python
df["Price"] = pd.to_numeric(df['Price'], errors = 'coerce')
```

```python
df['Price'].unique()
```

```python
df.info()
```

```python
df['kms_driven'].unique()
```

```python
df["kms_driven"] = (
    df["kms_driven"]
    .astype(str)
    .str.replace(r"[^\d]", "", regex=True)) 
```

```python
df['kms_driven'] = pd.to_numeric(df['kms_driven'], errors = 'coerce')
```

```python
df.isnull().sum() / len(df) * 100
```

```python
df = df.dropna(subset=["year","Price","kms_driven","fuel_type"])
```

```python
q1 = df["Price"].quantile(0.25)
q3 = df["Price"].quantile(0.75)
iqr = q3 - q1

upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

print("upper limit:", upper, "| lower limit:", lower)
print((df["Price"] > upper).sum(), "rows above upper")
print((df["Price"] < lower).sum(), "rows below lower")
```
